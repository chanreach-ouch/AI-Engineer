```python
import cohere
from groq import Groq
import os

groq_client = Groq(
    api_key=os.environ.get("GROQ_API_KEY"),
)
response = groq_client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": "Write a short poem about the beauty of nature."}], # Changed 'prompt' to 'contents'
)

print(response.choices[0].message.content) # Getting the response is also simplified now!
```

```python
import os
from groq import Groq

# Initialize the Groq client. 
# It will use the GROQ_API_KEY from your environment/ .env file
groq_client = Groq(
    api_key=os.environ.get("GROQ_API_KEY"),
)

response = groq_client.chat.completions.create(
    model="llama-3.1-8b-instant", # You can also try "mixtral-8x7b-32768" or "gemma2-9b-it"
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a short haiku about fast inference."},
    ]
)

print(response.choices[0].message.content)
```
